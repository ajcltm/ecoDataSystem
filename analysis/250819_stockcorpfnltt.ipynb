```python
import pandas as pd

valid_data = [d for d in staged if not "error" in d]
len(valid_data)
```

```python
valid_df = pd.DataFrame(valid_data)
valid_df.head()
```

```python
valid_df.columns
view_columns = ['corp_code', 'fs_div', '유동자산', '비유동자산', '자산총계', '부채총계', '자본총계', '매출액', '영업이익', '당기순이익', '당기순이익(손실)']
vv_df = valid_df[view_columns]
vv_df.head()
```

```python
vv_df.info()
```

```python
numeric_columns = ['유동자산', '비유동자산', '자산총계', '부채총계', '자본총계', '매출액', '영업이익', '당기순이익', '당기순이익(손실)']
vv_df.apply(lambda col: col.str.replace(',', '', regex=True) if col.name in numeric_columns else col).apply(lambda col: col.str.replace("-", "", regex=True) if col.name in numeric_columns else col).apply(pd.to_numeric, errors='coerce'  ).info()
vv_df.head()
```
