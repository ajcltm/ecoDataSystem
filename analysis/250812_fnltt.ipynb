```python
from pathlib import Path
import sys

def add_project_root_for_src():
    p = Path.cwd()
    for parent in [p, *p.parents]:
        if (parent / "src").is_dir():
            sys.path.append(str(parent))  # ← 'src'가 아닌, 그 부모를 추가해야 함
            return
    raise RuntimeError("상위 경로에서 'src' 폴더를 찾지 못했습니다.")

add_project_root_for_src()

# 이제 import가 됨
from src.etl.extract.Extracter_fnltt import Extracter_fnltt  # 이름/대소문자 확인!
```

```python
def extract_fnltt(data):
    if data.get("status") == "000":
        dic = dict()
        for item in data.get("list", []):
            key = item.get("account_nm")
            value = item.get("thstrm_amount")
            dic[key] = value
        return dic
```

```python
dic = extract_fnltt(data)
print(dic)
```

```python
def check_fs_div(data):
    for item in data.get("list", []):
        if item.get("fs_div") == "CFS":
            return "CFS"
    return "OFS"
```

```python
check_fs_div(data)
```

```python
def get_report(data, fs_div):
    if data.get("status") == "000":
        print(f"Financial Statements Division: {fs_div}")
        dic = dict()
        for item in data.get("list", []):
            if item.get("fs_div") == fs_div:
                key = item.get("account_nm")
                value = item.get("thstrm_amount")
                print(f"{key}: {value}")
                dic[key] = value
        return dic
```

```python
get_report(data, check_fs_div(data))
```
